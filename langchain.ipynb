```python
import os
import json
# Remove deprecated import
# from langchain import HuggingFaceHub
# Updated: Use langchain_huggingface instead of deprecated HuggingFaceHub
from langchain_huggingface import HuggingFaceEndpoint
from langchain_groq import ChatGroq

with open('configs/config.json') as f:
    config = json.load(f)
os.environ["GROQ_API_KEY"] = config['groq_api_key']
os.environ["HUGGINGFACEHUB_API_TOKEN"] = config['huggingfacehub_api_token']
```

temperature value --> hOW CREATIVE THE MODEL IS IN ITS RESPONSES. A HIGHER VALUE (LIKE 0.8) WILL MAKE THE OUTPUT MORE CREATIVE AND UNPREDICTABLE, WHILE A LOWER VALUE (LIKE 0.2) WILL MAKE IT MORE FOCUSED AND DETERMINISTIC.

```python
llm_groq=ChatGroq(
    model="openai/gpt-oss-120b", 
    temperature=0.6, 
    groq_api_key=config['groq_api_key']
    )
```

```python
llm_huggingFace = HuggingFaceEndpoint(
    repo_id="mistralai/Mistral-7B-Instruct-v0.3",  # free, hosted, text2text-generation
    temperature=0.6,
    task="conversational"
)
```

```python
# Use the correct method to invoke the HuggingFaceEndpoint LLM and get the output
result = llm_huggingFace.invoke("What is the capital of Russia")
# The result is a string, print directly
print(result)
```
